## Summary

This project explored whether batting average predicts runs scored in the 2026 MLB season.

- Batting Average (AVG) had a moderate correlation with runs (r = 0.64)
- On-Base Percentage (OBP) was a stronger predictor (r = 0.78)
- Slugging Percentage (SLG) was similarly strong (r = 0.79)
- OPS (OBP + SLG combined) was the strongest predictor of all (r = 0.88)

**Conclusion:** Getting on base and hitting for power together (OPS) explain run production far better than batting average alone — echoing the "Moneyball" philosophy of valuing OBP/SLG over AVG.

```python
from sklearn.linear_model import LinearRegression
```

```python
X = merged[['OPS']]
y = merged['W']

model = LinearRegression()
model.fit(X, y)
```

```python
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Regression Equation

Wins = (slope × OPS) + intercept
Wins = (108.79 × OPS) + (-25.51)

Meaning: each 0.01 increase in OPS ≈ 1.09 more wins

```python
r_squared = model.score(X, y)
print("R²:", r_squared)
```

## SUMMARY 
"A simple linear regression using only OPS explains about 17% of the variation in team wins (R² = 0.169). This confirms that while OPS is a meaningful offensive indicator, it is far from sufficient on its own to predict a team's win total — consistent with the idea that winning depends heavily on factors OPS doesn't capture, particularly pitching and run prevention."

```python
ra_corr = merged['RA'].corr(merged['W'])
rdiff_corr = merged['Rdiff'].corr(merged['W'])
```

```python
print("OPS vs Wins:", correlation_wins)
print("RA vs Wins:", ra_corr)
print("Rdiff vs Wins:", rdiff_corr)
```

## Choosing the Second Predictor: Why RA, not Rdiff

Three stats were compared against Wins:

| Stat | Correlation with Wins |
|---|---|
| Rdiff (Run Differential) | 0.91 |
| RA (Runs Allowed) | -0.71 |
| OPS | 0.41 |

Rdiff was excluded despite its very strong correlation, because 
Rdiff = Runs Scored − Runs Allowed, making it mathematically close to 
the definition of winning itself — using it as a predictor would be 
somewhat circular.

RA was chosen instead. Its correlation is negative, as expected 
(more runs allowed → fewer wins). Unlike Rdiff, RA captures a genuinely 
separate skill (pitching/run prevention) that OPS does not measure, 
making OPS + RA a combination of two independent signals rather than 
one restating the outcome.

```python
X = merged[['OPS', 'RA']]
y = merged['W']

OPS_RA_model = LinearRegression()
OPS_RA_model.fit(X, y)

print("Coefficients:", OPS_RA_model.coef_)
print("Intercept:", OPS_RA_model.intercept_)
```

```python
r_squared_OPS_RA = OPS_RA_model.score(X, y)
print("R² (OPS + RA):", r_squared_OPS_RA)
```

## CONFIRMATION
"A model using only OPS explained just 17% of the variation in team wins. Adding Runs Allowed (RA) as a second predictor — representing pitching and run prevention — dramatically improved the model to explain 76% of win variation. This confirms that winning in baseball depends on both offensive production and run prevention, and that OPS alone, while a strong predictor of scoring, is insufficient to explain overall team success."

```python
predicted_wins = OPS_RA_model.predict(X)
plt.scatter(y, predicted_wins)
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linestyle='--')
plt.xlabel('Actual Wins')
plt.ylabel('Predicted Wins')
plt.title('How Close Are Our Predictions to Reality?')
plt.show()
```

## Visualizing Model Accuracy: Actual vs Predicted Wins

The chart above compares each team's actual 2026 win total to what the OPS + RA model predicted for them. The red dashed line represents a perfect prediction — if a team's dot sits exactly on this line, the model guessed their win total exactly right.

Most teams cluster closely around the line, confirming the model's strong R² of 0.756. A few teams fall noticeably above or below it — these are cases where factors outside OPS and RA (such as bullpen performance, injuries, or close-game luck) likely influenced their actual win total more than their underlying stats would suggest.

```python
merged['predicted_wins'] = predicted_wins
merged['difference'] = merged['W'] - merged['predicted_wins']
merged[['Tm', 'W', 'predicted_wins', 'difference']].sort_values('difference')
```
